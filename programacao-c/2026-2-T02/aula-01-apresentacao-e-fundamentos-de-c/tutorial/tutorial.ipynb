## O programa da aula

Antes de modificar, recompile e execute o programa condutor **exatamente como saiu da aula** e confira que a saída bate com a que vimos em aula.

```python
%%writefile media_turma.c
#include <stdio.h>
#define TAM 5

double media(double v[], int n) {
    double soma = 0.0;
    for (int i = 0; i < n; i++)
        soma += v[i];
    return soma / n;
}

int main(void) {
    double notas[TAM];
    printf("Digite %d notas: ", TAM);
    for (int i = 0; i < TAM; i++)
        scanf("%lf", &notas[i]);
    double m = media(notas, TAM);
    printf("Media: %.2f\n", m);
    if (m >= 5.0)
        printf("Turma aprovada\n");
    else if (m >= 3.0)
        printf("Turma em reposicao\n");
    else
        printf("Turma reprovada\n");
    return 0;
}
```

## 2. Tipos de dados, variáveis e E/S

Os tipos básicos de C são `int`, `float`, `double`, `char` e `long`. O tamanho de cada tipo depende da plataforma — o operador `sizeof` revela quantos bytes cada um ocupa. Para ler valores do teclado usamos `scanf`, passando o **endereço** da variável com `&` (uma prévia de ponteiros, tema das próximas semanas).

```python
%%writefile tipos.c
#include <stdio.h>

int main(void) {
    printf("int:    %zu bytes\n", sizeof(int));
    printf("float:  %zu bytes\n", sizeof(float));
    printf("double: %zu bytes\n", sizeof(double));
    printf("char:   %zu byte\n",  sizeof(char));
    printf("long:   %zu bytes\n", sizeof(long));
    return 0;
}
```

```python
!gcc tipos.c -o tipos && ./tipos
```

```python
%%writefile media.c
#include <stdio.h>

int main(void) {
    double nota1, nota2;
    scanf("%lf %lf", &nota1, &nota2);   /* & = endereco da variavel */
    double media = (nota1 + nota2) / 2.0;
    printf("Media = %.2f\n", media);
    return 0;
}
```

## 3. Controle de fluxo: decisão

O `if/else if/else` encadeia condições avaliadas em ordem; o `switch` compara uma expressão inteira com valores fixos (`case`) — sem o `break`, a execução "vaza" para o próximo caso.

```python
%%writefile decisao.c
#include <stdio.h>

int main(void) {
    double media;
    scanf("%lf", &media);
    if (media >= 5.0)
        printf("Aprovado\n");
    else if (media >= 3.0)
        printf("Reposicao\n");
    else
        printf("Reprovado\n");
    return 0;
}
```

## 4. Controle de fluxo: repetição

Use `for` quando o número de repetições é conhecido, `while` quando depende de uma condição, e `do-while` quando o corpo deve executar ao menos uma vez (menus, validação de entrada).

```python
%%writefile lacos.c
#include <stdio.h>

int main(void) {
    for (int i = 0; i < 5; i++)          /* 0 1 2 3 4 */
        printf("%d ", i);
    printf("\n");

    int n = 32, passos = 0;
    while (n > 1) {                      /* quantas divisoes por 2? */
        n = n / 2;
        passos++;
    }
    printf("32 vira 1 em %d divisoes\n", passos);
    return 0;
}
```

## 5. Funções

Funções organizam o código em unidades reutilizáveis. O **protótipo** declara a assinatura antes do `main`; a **definição** traz o corpo. Em C os argumentos são **copiados** (passagem por valor) — alterar o parâmetro dentro da função não muda a variável do chamador. Para isso precisaremos de ponteiros.

```python
%%writefile funcoes.c
#include <stdio.h>

double media(double a, double b);        /* prototipo */

int main(void) {
    printf("%.2f\n", media(7.0, 9.0));
    return 0;
}

double media(double a, double b) {       /* definicao */
    return (a + b) / 2.0;
}
```

## 6. Vetores e strings

Um vetor ocupa posições **contíguas** na memória e seus índices vão de `0` a `tamanho - 1`. C **não verifica limites**: acessar `v[5]` em um vetor de 5 posições é um erro silencioso — fonte clássica de bugs. Strings são vetores de `char` terminados pelo caractere nulo `'\0'`.

```python
%%writefile vetores.c
#include <stdio.h>
#define TAM 5

int main(void) {
    double notas[TAM] = {7.5, 8.0, 6.5, 9.0, 5.5};
    double soma = 0.0;
    for (int i = 0; i < TAM; i++)
        soma += notas[i];
    printf("Media da turma: %.2f\n", soma / TAM);

    char nome[] = "COMP0512";            /* string = vetor de char */
    printf("Disciplina: %s\n", nome);
    return 0;
}
```
