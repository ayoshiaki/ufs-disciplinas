## O programa da aula

Antes de alterar, recompile e execute o programa **exatamente como saiu da aula** e confira que a saída bate com a que vimos em sala.

```python
%%writefile notas_turma.c
#include <stdio.h>
#define CAPACIDADE 8

int inserir(double v[], int n, double nota) {
    if (n == CAPACIDADE)
        return n;          /* colecao cheia: ignora a nota */
    v[n] = nota;
    return n + 1;
}

double media(double v[], int n) {
    double soma = 0.0;
    for (int i = 0; i < n; i++)
        soma += v[i];
    return soma / n;
}

double maior(double v[], int n) {
    double m = v[0];
    for (int i = 1; i < n; i++)
        if (v[i] > m)
            m = v[i];
    return m;
}

int main(void) {
    double notas[CAPACIDADE];
    int n = 0;

    n = inserir(notas, n, 7.0);
    n = inserir(notas, n, 8.5);
    n = inserir(notas, n, 6.0);

    printf("%d notas cadastradas\n", n);
    printf("media: %.2f\n", media(notas, n));
    printf("maior: %.1f\n", maior(notas, n));
    printf("cada nota ocupa %zu bytes\n", sizeof(double));
    printf("o vetor ocupa %zu bytes\n", sizeof notas);
    return 0;
}
```

```python
# Compila e executa — a saída deve ser idêntica à vista em aula
!gcc -Wall notas_turma.c -o notas_turma && ./notas_turma
!./notas_turma | grep -q 'media: 7.17' && echo OK || echo 'Verifique: esperava media: 7.17'
```

### Prática 3 (abstração) — o cofrinho

Implemente um TAD **cofrinho** com duas operações sobre um saldo:

- `double depositar(double saldo, double valor)` — devolve o novo saldo; ignora valores negativos;
- `double sacar(double saldo, double valor)` — devolve o novo saldo; **não permite** sacar mais do que há no cofrinho (nesse caso devolve o saldo inalterado).

As regras vivem **dentro** das operações: quem usa o cofrinho não consegue deixá-lo negativo.

```python
%%writefile cofrinho.c
#include <stdio.h>

/* TODO: implemente depositar e sacar */

int main(void) {
    double saldo = 0.0;
    saldo = depositar(saldo, 50.0);
    saldo = depositar(saldo, -10.0);  /* deve ser ignorado */
    saldo = sacar(saldo, 20.0);
    saldo = sacar(saldo, 100.0);      /* deve ser recusado */
    printf("saldo final: %.2f\n", saldo);
    return 0;
}
```
