## De onde vem este problema

> **Triagem de Risco.** A triagem hospitalar classifica cada paciente com base em dois
> dados: **pressão sistólica** (real, na escala mmHg/10 — o valor `14.0` representa
> 140 mmHg) e **idade** (inteiro, em anos). Cada paciente recebe **exatamente uma** das
> classificações abaixo:
>
> | Classificação | Descrição |
> |---|---|
> | `NORMAL`  | pressão e idade dentro dos limites saudáveis, sem nenhum fator de atenção |
> | `BAIXO`   | pressão levemente elevada, acima de 120 mmHg, sem outros agravantes |
> | `MEDIO`   | pressão elevada (acima de 140 mmHg) **ou** paciente idoso (mais de 60 anos) |
> | `ALTO`    | pressão elevada (acima de 140 mmHg) **e** paciente idoso (mais de 60 anos) |
> | `CRITICO` | pressão muito abaixo do normal, inferior a 90 mmHg — emergência, qualquer idade |
>
> *Atenção: um mesmo paciente pode se enquadrar em mais de uma descrição; cabe a você
> determinar a ordem correta de verificação.*

## 1. Antes de rodar: escreva a sua previsão

Abaixo está uma função pronta que resolve a triagem. **Não execute ainda.**

Leia o código com calma e preencha a tabela abaixo no papel (ou aqui mesmo, editando
esta célula). Só depois execute a célula seguinte.

| Paciente | Pressão | Idade | Minha previsão |

| Antonio | 15.0 | 70 | |
| Beatriz | 14.5 | 30 | |
| Cesar   | 12.5 | 80 | |

| Eduardo |  8.5 | 45 | |

> Errar a previsão não é problema — é exatamente onde o aprendizado acontece.
> O que importa é **anotar antes**.


````python
def risco(pressao, idade):
    if pressao < 9.0:
        return "CRITICO"
    elif pressao > 14.0 and idade > 60:
        return "ALTO"
    elif pressao > 14.0 or idade > 60:
        return "MEDIO"
    elif pressao > 12.0:
        return "BAIXO"
    else:
        return "NORMAL"


PACIENTES = [
    ("Antonio", 15.0, 70),
    ("Beatriz", 14.5, 30),
    ("Cesar",   12.5, 80),
    ("Daniela", 11.0, 20),
    ("Eduardo",  8.5, 45),
]

for nome, pressao, idade in PACIENTES:
    print(nome + ": " + risco(pressao, idade))
````

## 2. Investigando: o que o Python faz, linha por linha

### 2.1 Uma condição é uma pergunta de sim ou não

O `if` não entende "pressão alta". Ele só sabe olhar para um valor `True` ou `False`.
Execute a célula e observe o tipo de cada resultado.


````python
pressao = 15.0
idade = 70

print("pressao < 9.0        ->", pressao < 9.0)
print("pressao > 14.0       ->", pressao > 14.0)
print("idade > 60           ->", idade > 60)
print("tipo do resultado    ->", type(pressao > 14.0))

# Os operadores de comparacao disponiveis:
#   <   menor que          <=  menor ou igual a
#   >   maior que          >=  maior ou igual a
#   ==  e igual a          !=  e diferente de
print("idade == 70          ->", idade == 70)
print("idade != 70          ->", idade != 70)
````

### 2.2 `=` guarda, `==` pergunta

`idade = 70` **guarda** o valor 70 na variável.
`idade == 70` **pergunta** se a variável vale 70.

Trocar um pelo outro dentro de um `if` é erro de sintaxe — o Python nem executa o
programa. Descomente a linha abaixo para ver a mensagem de erro, depois comente de novo.


````python
# Descomente as duas linhas seguintes para ver o erro:
# if idade = 70:
#     print("idoso")

# A forma correta:
if idade == 70:
    print("a variavel idade vale mesmo 70")
````

### 2.3 `elif` faz uma peneira

Com `elif`, o Python testa de cima para baixo e **para no primeiro teste verdadeiro**.
Os testes de baixo nem chegam a ser executados.

A célula abaixo é a mesma função, só que falando em voz alta o que está avaliando.
Rode com o Antonio e acompanhe onde ela para.


````python
def risco_verboso(pressao, idade):
    print("  teste 1: " + str(pressao) + " < 9.0 ->", pressao < 9.0)
    if pressao < 9.0:
        return "CRITICO"

    print("  teste 2: " + str(pressao) + " > 14.0 and " + str(idade) + " > 60 ->",
          pressao > 14.0 and idade > 60)
    if pressao > 14.0 and idade > 60:
        return "ALTO"

    print("  teste 3: " + str(pressao) + " > 14.0 or " + str(idade) + " > 60 ->",
          pressao > 14.0 or idade > 60)
    if pressao > 14.0 or idade > 60:
        return "MEDIO"

    print("  teste 4: " + str(pressao) + " > 12.0 ->", pressao > 12.0)
    if pressao > 12.0:
        return "BAIXO"

    print("  nenhum teste deu True: cai no else")
    return "NORMAL"


for nome, pressao, idade in PACIENTES:
    print(nome + " (" + str(pressao) + ", " + str(idade) + "):")
    resultado = risco_verboso(pressao, idade)
    print("  -> " + resultado)
    print()
````

**Observe o Antonio:** o teste 2 deu `True` e a função retornou ali mesmo. Os testes
3 e 4 — que também dariam `True` — nunca foram avaliados.

É isso que o `elif` garante: **exatamente uma** alternativa é escolhida, e é sempre a
**primeira** verdadeira.

## 3. A ordem importa

### 3.1 Experimento: e se trocarmos `elif` por `if`?

As regras abaixo são as mesmas. Os números são os mesmos. A ordem é a mesma.
Só que agora são quatro `if` independentes, guardando o resultado numa variável.

**Preveja antes de rodar:** o que sai para o Antonio?


````python
def risco_if_solto(pressao, idade):
    classificacao = "NORMAL"
    if pressao < 9.0:
        classificacao = "CRITICO"
    if pressao > 14.0 and idade > 60:
        classificacao = "ALTO"
    if pressao > 14.0 or idade > 60:
        classificacao = "MEDIO"
    if pressao > 12.0:
        classificacao = "BAIXO"
    return classificacao


print("nome       correto    if solto")
for nome, pressao, idade in PACIENTES:
    certo = risco(pressao, idade)
    quebrado = risco_if_solto(pressao, idade)
    marca = "   " if certo == quebrado else " <-- ERRADO"
    print(nome.ljust(10), certo.ljust(10), quebrado + marca)
````

**Por que o Antonio virou `BAIXO`?**

Sem `elif` não há peneira: **todos** os testes são executados, e cada acerto sobrescreve
o anterior.

| # | Teste | Resultado | `classificacao` passa a valer |
|---|---|---|---|
|   | (valor inicial) | — | `"NORMAL"` |
| 1 | `15.0 < 9.0` | `False` | `"NORMAL"` |
| 2 | `15.0 > 14.0 and 70 > 60` | `True` | `"ALTO"` |
| 3 | `15.0 > 14.0 or 70 > 60` | `True` | `"MEDIO"` |
| 4 | `15.0 > 12.0` | `True` | `"BAIXO"` |

> Com `if` independentes, quem manda é o **último** teste verdadeiro.
> Com `elif`, quem manda é o **primeiro**.


### 3.2 Experimento: manter o `elif`, trocar a ordem

Agora a peneira está de volta — mas os testes estão em outra ordem.

**Preveja antes de rodar.**


````python
def risco_ordem_errada(pressao, idade):
    if pressao > 12.0:
        return "BAIXO"
    elif pressao > 14.0 or idade > 60:
        return "MEDIO"
    elif pressao > 14.0 and idade > 60:
        return "ALTO"
    elif pressao < 9.0:
        return "CRITICO"
    else:
        return "NORMAL"


print("nome       correto    ordem trocada")
for nome, pressao, idade in PACIENTES:
    certo = risco(pressao, idade)
    quebrado = risco_ordem_errada(pressao, idade)
    marca = "   " if certo == quebrado else " <-- ERRADO"
    print(nome.ljust(10), certo.ljust(10), quebrado + marca)
````

Dois erros **diferentes** produziram a **mesma** saída errada. Nos dois casos, o rótulo
mais genérico ganhou do mais específico.

### 3.3 A regra que resolve

As categorias se contêm umas às outras: todo paciente `ALTO` também satisfaz a
descrição de `MEDIO`, e todo `MEDIO` por pressão também satisfaz a de `BAIXO`.

Quando uma condição é um **caso particular** de outra, o caso particular tem de ser
testado **primeiro**. Daí a ordem correta:

1. `CRITICO` — emergência: vale sozinho, ignora tudo o mais
2. `ALTO` — dois agravantes ao mesmo tempo (`and`)
3. `MEDIO` — um agravante qualquer (`or`)
4. `BAIXO` — só um sinal leve
5. `NORMAL` — o que sobrou (`else`)

**Do mais específico para o mais geral.**

## 4. `and`, `or`, `not`

`ALTO` pede pressão elevada **e** paciente idoso → `and`.
`MEDIO` pede pressão elevada **ou** paciente idoso → `or`.

Rode a célula para ver a tabela verdade completa.


````python
print("   A       B     A and B   A or B")
for a in [True, False]:
    for b in [True, False]:
        print(str(a).ljust(7), str(b).ljust(6),
              str(a and b).ljust(9), str(a or b))

print()
print("not True  ->", not True)
print("not False ->", not False)

# ARMADILHA: cada lado do "and" precisa ser uma condicao completa.
# Isto NAO existe em Python:   if pressao > 14.0 and > 60:
# O correto e:                 if pressao > 14.0 and idade > 60:
````

## 5. Cuidado com as fronteiras

"Acima de 140" é `> 14.0`. **Não** é `>= 14.0`. Um caractere muda a resposta.

É exatamente em cima do limite que o erro se esconde — e é lá que a correção da prova
vai testar.


````python
LIMITES = [
    (14.0, 70, "MEDIO",   "140 nao e acima de 140; mas 70 e acima de 60"),
    (14.0, 30, "BAIXO",   "nenhum agravante; 140 esta acima de 120"),
    (12.0, 20, "NORMAL",  "120 nao e acima de 120"),
    (9.0,  45, "NORMAL",  "90 nao e inferior a 90"),
    (8.99, 45, "CRITICO", "agora sim, abaixo de 90"),
]

for pressao, idade, esperado, motivo in LIMITES:
    obtido = risco(pressao, idade)
    status = "ok" if obtido == esperado else "FALHOU"
    print(str(pressao).ljust(6), str(idade).ljust(4), obtido.ljust(8),
          status.ljust(7), motivo)
````
